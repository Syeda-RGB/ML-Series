# Data Pre-processing

```python
# Define X and y

X = df.drop (columns=['target'])
Y = df['target']
```

```python
# Split Data before scaling to prevent data leakage
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split( X,Y,test_size=0.2,random_state=42,stratify=Y)

#Preserve unscaled versions
X_train_unscaled = X_train.copy()
X_test_unscaled = X_test.copy()
```

```python
#Scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Models

# Logistic Regression ---> Scaled Data

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

model = LogisticRegression()

model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("Classification Report:")
print(classification_report(y_test, y_pred))
```

# Naive Bayes ---> unscaled data

```python
from sklearn.naive_bayes import GaussianNB

nb_model = GaussianNB()

nb_model.fit(X_train_unscaled, y_train)

y_pred_nb = nb_model.predict(X_test_unscaled)

print("Accuracy:", accuracy_score(y_test, y_pred_nb))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_nb))

print("Classification Report:")
print(classification_report(y_test, y_pred_nb))
```

# SVM ---> scaled data

```python
from sklearn.svm import SVC

svm_model = SVC()

svm_model.fit(X_train_scaled, y_train)

y_pred_svm = svm_model.predict(X_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred_svm))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_svm))

print("Classification Report:")
print(classification_report(y_test, y_pred_svm))
```

# KNN ---> scaled data

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier()

knn_model.fit(X_train_scaled, y_train)

y_pred_knn = knn_model.predict(X_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred_knn))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_knn))

print("Classification Report:")
print(classification_report(y_test, y_pred_knn))
```

# Decision Tree ---> unscaled data

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train_unscaled, y_train)

y_pred_dt = dt_model.predict(X_test_unscaled)

print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))

print("Classification Report:")
print(classification_report(y_test, y_pred_dt))
```

# Cross-Validation

```python
from sklearn.model_selection import cross_val_score
from sklearn.metrics import make_scorer, recall_score

malignant_recall = make_scorer(recall_score, pos_label=0)

# Decision Tree
scores = cross_val_score(dt_model,X_train_unscaled,y_train,cv=5,scoring=malignant_recall)

print(scores)
print("Mean Malignant Recall:", scores.mean())
```

```python
# Knn
scores = cross_val_score(knn_model,X_train_scaled,y_train,cv=5,scoring=malignant_recall)
print(scores)
print("Mean Malignant Recall:", scores.mean())
```

```python
# SVM
scores = cross_val_score(svm_model,X_train_scaled,y_train,cv=5,scoring=malignant_recall)

print(scores)
print("Mean Malignant Recall:", scores.mean())
```

```python
# Naive Baise
scores = cross_val_score(nb_model,X_train_unscaled,y_train,cv=5,scoring=malignant_recall)

print(scores)
print("Mean Malignant Recall:", scores.mean())
```

```python
#Logistic Regression
scores = cross_val_score(model,X_train_scaled,y_train,cv=5,scoring=malignant_recall)

print(scores)
print("Mean Malignant Recall:", scores.mean())
```

```python
### Cross-Validation Insights

#5-fold cross-validation was used to evaluate the models more reliably using
# Malignant Recall (class 0) as the main metric.

#- Logistic Regression achieved the highest mean CV recall (~96.5%), showing strong and consistent malignant-case detection.
#- SVM also performed strongly with a mean recall of ~95.9%.
#- KNN remained stable at ~93.5%, close to its original test-set recall.
#- Naive Bayes achieved ~89.4% recall and showed relatively consistent performance.
#- Decision Tree showed the largest decrease, from ~93% recall on the original test split to ~87.6% mean CV recall, indicating greater sensitivity to different data splits.

#Overall, cross-validation showed that a high score from a single train-test split does not always represent the model's general performance.
```

# Hyper Parameter Tunning

```python
#parameter options

param_grid = {
    'max_depth': [3, 5, 7, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 5],
    'criterion': ['gini', 'entropy']
}

#Apply GridSearchCV

from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(
    estimator=dt_model,
    param_grid=param_grid,
    cv=5,
    scoring=malignant_recall,
    n_jobs=-1
)

grid_search.fit(X_train_unscaled, y_train)

print("Best Parameters:")
print(grid_search.best_params_)

print("Best CV Malignant Recall:")
print(grid_search.best_score_)

best_dt = grid_search.best_estimator_

y_pred_tuned_dt = best_dt.predict(X_test_unscaled)

print("Accuracy:", accuracy_score(y_test, y_pred_tuned_dt))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_tuned_dt))

print("Classification Report:")
print(classification_report(y_test, y_pred_tuned_dt))
```

```python
### Hyperparameter Tuning Insight

#GridSearchCV was applied to the Decision Tree using 5-fold cross-validation
#and malignant recall as the scoring metric.

#Best parameters:
#- criterion = entropy
#- max_depth = 5
#- min_samples_split = 2
#- min_samples_leaf = 2

#The mean cross-validation malignant recall improved from approximately
#87.6% to 92.9%.

#On the final test set, malignant recall improved from 93% to 95%,
#while overall accuracy increased from approximately 91% to 93%.

#This shows that hyperparameter tuning improved both the model's
#generalization across folds and its ability to detect malignant cases.
```
